A100 GPU Optimization Commands for FaceFusion
This document contains optimized commands for running FaceFusion on an A100 GPU, designed to maximize both quality and performance.

###Baseline A100 GPU Command
python facefusion.py headless-run \
--keep-temp \
--log-level debug \
--execution-providers cuda \
--execution-thread-count 16 \
--execution-queue-count 2 \
--video-memory-strategy tolerant \
--temp-frame-format jpg \
--face-selector-mode reference \
--face-mask-types box occlusion \
--processors face_swapper \
--face-swapper-model inswapper_128 \
--face-swapper-pixel-boost 512x512 \
--face-detector-model retinaface \
--face-detector-size 640x640 \
--face-detector-score 0.5 \
--reference-face-distance 0.6 \
--face-mask-blur 0.5 \
--output-video-preset veryfast \
--output-video-quality 85 \
-s images/image.png \
-t images/videos/vid5.mp4 \
-o images/results/command-videos/baseline_result_vid5.mp4

This baseline command leverages the A100's power with CUDA acceleration while maintaining high-quality face swaps.

-o images/results/result_vid1.mp4

This command uses the higher-quality Ghost model and adds multiple face detection angles for difficult videos.
